```python
#PRINCIPAL COMPONENT ANALYSIS
#Subset numerical data

Churn_clean_data = Churn_data_df[['Population', 'Children', 'Age','Income' , 'Outage_sec_perweek', 'Email', 'Contacts', 'Yearly_equip_failure', 'Tenure', 'MonthlyCharge', 'Bandwidth_GB_Year', 'item1', 'item2', 'item3', 'item4', 'item5', 'item6', 'item7', 'item8']]

Churn_clean_data
```

```python
#calculating Principal Components
import sklearn as sk
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
std_df=scaler.fit_transform(Churn_clean_data)
from sklearn.decomposition import PCA
pca=PCA()
print(pca.fit_transform(std_df))
```

```python
#Calculating principal component explained variance ratio
pca.fit(std_df)
print(pca.explained_variance_ratio_)
```

```python
#Utilizing PCA for dimensionality reduction
print(pca.explained_variance_ratio_.cumsum())
```

```python
print(pca.components_)
```

```python
# Normalize the data
data=Churn_clean_data
churn_normalized = (data - data.mean()) / data.std()
```

```python
# Select number of components to extract
pca = PCA(n_components = data.shape[1])
```

```python
 # Create a list of PCA names
churn_numeric = data[['Population', 'Children', 'Age','Income' , 'Outage_sec_perweek', 'Email', 'Contacts', 'Yearly_equip_failure', 'Tenure', 'MonthlyCharge', 'Bandwidth_GB_Year', 'item1', 'item2', 'item3', 'item4', 'item5', 'item6', 'item7', 'item8']]
PCA_names = []
for i, col in enumerate(churn_numeric.columns):PCA_names.append('PCA' + str(i + 1))
print(PCA_names)
```

```python
# Call PCA application and replace the original dataset of 19 variables into a dataset of 19 components
pca.fit(churn_normalized)
churn_pca = pd.DataFrame(pca.transform(churn_normalized),
columns = PCA_names)
```

```python
# Create the scree plot
plt.plot(pca.explained_variance_ratio_)
plt.xlabel('Number of Components')
plt.ylabel('Explained Variance')
plt.show();
```

```python
# Extract the eigenvalues or proper values from analysis
cov_matrix = np.dot(churn_normalized.T, churn_normalized) / data.shape[0]
eigenvalues = [np.dot(eigenvector.T, np.dot(cov_matrix, eigenvector)) for eigenvector in pca.components_]
```

```python
# Plot the eigenvalues
plt.plot(eigenvalues)
plt.xlabel('Number of Components')
plt.ylabel('eigenvalues')
plt.show();
```

```python
 # Select the top components that explain variance
for pc, var in zip(PCA_names, np.cumsum(pca.explained_variance_ratio_)):print(pc, var)
```

```python
# Above, we see that 86% of variance is explained by 13 components
# Create a rotation
rotation = pd.DataFrame(pca.components_.T, columns = PCA_names, index = churn_numeric.columns)
print(rotation)
```

```python
# Output Churn_data_loadings for components
Churn_data_loadings = pd.DataFrame(pca.components_.T,
columns = PCA_names,
index = data.columns)
Churn_data_loadings
```
