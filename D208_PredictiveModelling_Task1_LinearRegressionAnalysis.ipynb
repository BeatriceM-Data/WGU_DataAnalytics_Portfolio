```python
categorical_columns = Churn_df.select_dtypes(include=['object']).columns
def describe_dataframe_type(Churn_df, columns):
    """
    Describe categorical and boolean columns in a DataFrame.
    
    Parameters:
        Churn_df (pd.DataFrame): The DataFrame to analyze.
        columns (list): List of columns to describe.
    
    Returns:
        pd.DataFrame: Summary of categorical and boolean features.
    """
    summary = {}
    for col in columns:
        if col in Churn_df.columns:
            unique_vals = Churn_df[col].nunique()
            values = Churn_df[col].unique()
            top_val = Churn_df[col].mode()[0]
            freq = Churn_df[col].value_counts().iloc[0]
            missing = Churn_df[col].isnull().sum()
            
            summary[col] = {
                'Data type': 'Categorical',
                
                'No. of Unique values': unique_vals,
                'Top value':top_val,
                'Frequency of the top value': freq,
                'The missing values': missing,
                'The unique Values are ': values
            }
    
    return pd.DataFrame(summary).T

# Usage example
description = describe_dataframe_type(Churn_df, categorical_columns)
description.head(40)

```

```python
print("INITIAL MULTIPLE LINEAR REGRESSION MODEL")

#The model will include all the predictor variables (multiple regression)which will be reduced using the p_values 
#so as to obtain the most valuable features that contribute in determining the monthly charge.

# convert array to list then append lists then remove target
numerical_features = Churn_df_encoded.select_dtypes(include="number").columns
model_features = numerical_features.tolist()

target = 'MonthlyCharge'
model_features.remove(target)
y = Churn_df_encoded[target]
X = Churn_df_encoded[model_features]
X1=X.astype(np.float64)
# initial model
Xc = sm.tools.tools.add_constant(X1)

Churn_df_encoded_model = lm.OLS(y, Xc).fit()
print(Churn_df_encoded_model.summary2()) # using alternate summary layout
```

```python
def mse(actual, pred):
    actual, pred = np.array(actual), np.array(pred)
    return np.square(np.subtract(actual, pred)).mean()

def rmse(actual, pred):
    return np.sqrt(mse(actual, pred))

# Assuming `y` is the actual target variable (e.g., MonthlyCharge)
# and `Churn_df_encoded_model` is the fitted model from statsmodels

pred = Churn_df_encoded_model.predict(Xc)  

# Calculate MSE and RMSE
mse_value = mse(y, pred)
rmse_value = rmse(y, pred)

# Print the results
print(f"MSE: {mse_value}")
print(f"RMSE: {rmse_value}")
```

```python
print("MODEL REDUCTION")
#Feature reduction using p_value
# drop all columns from model where p-value > 0.05 
equation = Churn_df_encoded_model.summary2().tables[1]
temp_drop = []
for i in equation.itertuples():
    if i[4] > 0.05:
        temp_drop.append(i[0])
        print('Drop {} with p-value of {:.3f}.'.format(i[0],i[4]))
X = pd.DataFrame(X) # reset dataframe
X.drop(temp_drop, axis = 1, inplace=True) # drop
```

```python
print("Update Model 1")
X.info()

# updated model
Xc = sm.tools.tools.add_constant(X)
Churn_df_encoded_model = lm.OLS(y, Xc).fit()
print(Churn_df_encoded_model.summary2()) 

```

```python
# drop all columns from model where p-value > 0.05 
equation = Churn_df_encoded_model.summary2().tables[1]
temp_drop = []
for i in equation.itertuples():
    if i[4] > 0.05:
        temp_drop.append(i[0])
        print('Drop {} with p-value of {:.3f}.'.format(i[0],i[4]))
X = pd.DataFrame(X) # reset dataframe
valid_drop = [col for col in temp_drop if col in X.columns]
X.drop(valid_drop, axis=1, inplace=True)  # Drop valid columns
#X.drop(temp_drop, axis = 1, inplace=True) # drop
print(f"Columns dropped: {valid_drop}")
```

```python
print("REDUCED LINEAR REGRESSION MODEL")
print("FINAL MODEL")
X.info()

# updated model
Xc = sm.tools.tools.add_constant(X)
Churn_df_encoded_model = lm.OLS(y, Xc).fit()
print(Churn_df_encoded_model.summary2()) 
```
